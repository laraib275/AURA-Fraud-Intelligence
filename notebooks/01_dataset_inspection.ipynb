```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/raw/sparkov/fraudTrain.csv")

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Remove the original CSV index column.
df = df.drop(columns=["Unnamed: 0"])

# Convert timestamp once and reuse it.
df["transaction_time"] = pd.to_datetime(df["trans_date_trans_time"])
```

```python
df.info()
```

```python
print("Missing values:")
display(df.isna().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Number of columns:", len(df.columns))

for i, col in enumerate(df.columns):
    print(i, "->", col)
```

```python
fraud_counts = df["is_fraud"].value_counts()

print(fraud_counts)
print("\nPercentage:")
print(df["is_fraud"].value_counts(normalize=True) * 100)
```

```python
print("Unique customers:", df["cc_num"].nunique())
print("Unique merchants:", df["merchant"].nunique())
print("Unique categories:", df["category"].nunique())
print("Unique cities:", df["city"].nunique())
print("Unique states:", df["state"].nunique())
print("Unique transaction IDs:", df["trans_num"].nunique())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate transaction IDs:", df["trans_num"].duplicated().sum())
```

```python
df["transaction_time"] = pd.to_datetime(df["trans_date_trans_time"])

print("Start:", df["transaction_time"].min())
print("End:", df["transaction_time"].max())
print("Number of days:",
      (df["transaction_time"].max() - df["transaction_time"].min()).days)
```

```python
print(df["amt"].describe())
print("Minimum:", df["amt"].min())
print("Maximum:", df["amt"].max())
print("Median:", df["amt"].median())
```

```python
df.groupby("is_fraud")["amt"].describe()
```

```python
fraud_by_category = (
    df.groupby("category")["is_fraud"]
      .agg(
          transaction_count="count",
          fraud_count="sum",
          fraud_rate="mean"
      )
      .sort_values("fraud_rate", ascending=False)
)

fraud_by_category
```

```python
customer_transaction_counts = (
    df.groupby("cc_num")["trans_num"]
      .count()
      .sort_values(ascending=False)
)

print("Customers:", customer_transaction_counts.shape[0])
print("Min transactions/customer:", customer_transaction_counts.min())
print("Max transactions/customer:", customer_transaction_counts.max())
print("Median transactions/customer:", customer_transaction_counts.median())

print("\nDistribution:")
display(customer_transaction_counts.describe())
```

```python
customer_transaction_counts = (
    df.groupby("cc_num")["trans_num"]
      .count()
      .sort_values(ascending=False)
)

print("Customers:", customer_transaction_counts.shape[0])
print("Min transactions/customer:", customer_transaction_counts.min())
print("Max transactions/customer:", customer_transaction_counts.max())
print("Median transactions/customer:", customer_transaction_counts.median())

print("\nDistribution:")
display(customer_transaction_counts.describe())
```
